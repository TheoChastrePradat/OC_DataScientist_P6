# Projet 6 - Classifier automatiquement des biens de consommation

## Ici nous allons tester une API pour collecter des donnees depuis Open Food Fact

### L'idee est d'extraire 10 premiers produits a base de champagne et de les enregistrer dans un csv. Pour chaque produit il faudra les informations suivantes : foodId, label, category, foodContentsLabel, image

### Commencons par enregistrer les champs obligatoires pour chaque produit qui seront necessaires lors du GET

```python
import requests
import pandas as pd
```

```python
FIELDS = "product_name,code,labels,categories,ingredients_text,selected_images"
```

```python
URL = ("https://world.openfoodfacts.org/cgi/search.pl"
       "?search_terms=champagne" # recherche de produits contenant "champagne"
       "&page_size=10" # nombre de résultats par page
       "&json=1" # format de réponse JSON
       f"&fields={FIELDS}") # les champs à récupérer
```

### Creons notre requete

```python
response = requests.get(URL).json()
```

```python
response
```

### Et enfin ici nous allons stocker les informations attendues pour chaque produit que nous avons recupere depuis notre requete

```python
data = []
for p in response['products']:
    data.append({
        "foodId": p.get("code"),
        "label": p.get("product_name"),
        "category": p.get("categories"),
        "foodContentsLabel": p.get("ingredients_text"),
        "image": p.get("selected_images", {}).get("front", {}).get("display")
    })
```

```python
data_champagne = pd.DataFrame(data)
data_champagne
```

### Et enfin notre exportation csv

```python
pd.DataFrame(data).to_csv("champagne_products.csv", index=False)
```
